```python
import time
import numpy as np
import matplotlib.pyplot as plt


def matmul(A, B):
    n = len(A)
    C = [[0.0] * n for _ in range(n)]
    for i in range(n):
        for j in range(n):
            s = 0.0
            for k in range(n):
                s += A[i][k] * B[k][j]
            C[i][j] = s
    return C


sizes = [10, 20, 40, 80, 120]
t_list, t_np = [], []

for n in sizes:
    A = np.random.rand(n, n)
    B = np.random.rand(n, n)
    Al, Bl = A.tolist(), B.tolist()

    t0 = time.perf_counter(); matmul(Al, Bl); t_list.append(time.perf_counter() - t0)
    t0 = time.perf_counter(); A @ B;          t_np.append(time.perf_counter() - t0)

    print(f"n={n:3d}  lists={t_list[-1]:.4f}s  numpy={t_np[-1]:.6f}s")

plt.loglog(sizes, t_list, 'o-', label='lists')
plt.loglog(sizes, t_np, 's-', label='numpy')
plt.xlabel('n'); plt.ylabel('time [s]'); plt.legend(); plt.grid(True, which='both')
plt.show()
```

```python
"""
Donut dataset with quadratic feature expansion.

Two concentric annular clouds:
    y = 0  -> inner ring  (R = 5)
    y = 1  -> outer ring  (R = 10)

Both drawn as R_mean + N(0, sigma) in the radial direction, uniform in angle.
The dataframe is built first with (x1, x2, y), then expanded with
x1^2, x2^2, x1*x2 so that column order matches the spec:

    x1, x2, x1^2, x2^2, x1*x2, y
"""

import numpy as np
import pandas as pd
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

N_PER_CLASS = 3000
R_INNER, R_OUTER = 5.0, 10.0
SIGMA = 1.5
SEED = 0

CSV_PATH = "/mnt/user-data/outputs/donut.csv"
PNG_PATH = "/mnt/user-data/outputs/donut.png"


def sample_ring(n, radius, sigma, rng):
    """n points scattered around a circle of given radius."""
    theta = rng.uniform(0.0, 2.0 * np.pi, n)
    r = radius + rng.normal(0.0, sigma, n)
    return np.column_stack([r * np.cos(theta), r * np.sin(theta)])


def make_donut(n_per_class=N_PER_CLASS, seed=SEED):
    rng = np.random.default_rng(seed)
    inner = sample_ring(n_per_class, R_INNER, SIGMA, rng)
    outer = sample_ring(n_per_class, R_OUTER, SIGMA, rng)

    X = np.vstack([inner, outer])
    y = np.concatenate([np.zeros(n_per_class, dtype=int),
                        np.ones(n_per_class, dtype=int)])

    # Shuffle so the CSV is not sorted by class.
    idx = rng.permutation(X.shape[0])
    X, y = X[idx], y[idx]

    return pd.DataFrame({"x1": X[:, 0], "x2": X[:, 1], "y": y})


def add_quadratic_features(df):
    """Insert x1^2, x2^2, x1*x2 between the coordinates and the label."""
    df = df.copy()
    df["x1^2"] = df["x1"] ** 2
    df["x2^2"] = df["x2"] ** 2
    df["x1*x2"] = df["x1"] * df["x2"]
    return df[["x1", "x2", "x1^2", "x2^2", "x1*x2", "y"]]


def plot_donut(df, path):
    fig, ax = plt.subplots(figsize=(6, 6))
    ax.scatter(df["x1"], df["x2"], c=df["y"], s=8, cmap="viridis", alpha=0.8)
    ax.set_aspect("equal")
    ax.set_xlabel("$x_1$")
    ax.set_ylabel("$x_2$")
    ax.set_title("Donut dataset")
    fig.tight_layout()
    fig.savefig(path, dpi=150)
    plt.close(fig)


if __name__ == "__main__":
    df = add_quadratic_features(make_donut())
    plot_donut(df, PNG_PATH)
    df.to_csv(CSV_PATH, header=False, index=False)

    print(df.head())
    print(df.shape, df["y"].value_counts().to_dict())
```

```python
"""
Sobel edge detection.

    Gx = Hx * A,  Gy = Hy * A,  G = sqrt(Gx^2 + Gy^2)

A is the grayscale image, * is 2D convolution.
"""

import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import convolve2d

Hx = np.array([[1, 0, -1],
               [2, 0, -2],
               [1, 0, -1]], dtype=float)

Hy = np.array([[ 1,  2,  1],
               [ 0,  0,  0],
               [-1, -2, -1]], dtype=float)


def sobel(gray):
    """Gradient magnitude of a 2D float array."""
    Gx = convolve2d(gray, Hx, mode='same', boundary='symm')
    Gy = convolve2d(gray, Hy, mode='same', boundary='symm')
    return np.sqrt(Gx**2 + Gy**2), Gx, Gy


if __name__ == "__main__":
    from scipy.datasets import ascent          # any 2D array works
    gray = ascent().astype(float)

    G, Gx, Gy = sobel(gray)

    fig, axes = plt.subplots(1, 4, figsize=(16, 4))
    for ax, img, title in zip(axes,
                              [gray, Gx, Gy, G],
                              ['A', 'Gx (vertical edges)',
                               'Gy (horizontal edges)', 'G']):
        # Gx, Gy are signed -> symmetric range around 0; G is non-negative.
        if title.startswith('G') and 'G ' not in title and title != 'G':
            lim = np.abs(img).max()
            ax.imshow(img, cmap='gray', vmin=-lim, vmax=lim)
        else:
            ax.imshow(img, cmap='gray')
        ax.set_title(title)
        ax.axis('off')
    fig.tight_layout()
    plt.show()
```
